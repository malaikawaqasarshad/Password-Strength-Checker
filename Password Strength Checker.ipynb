```python
has_upper = any(char.isupper() for char in password)
has_lower = any(char.islower() for char in password)

if has_upper:
    print("✓ Contains an uppercase letter")
else:
    print("✗ Missing an uppercase letter")

if has_lower:
    print("✓ Contains a lowercase letter")
else:
    print("✗ Missing a lowercase letter")
```

```python
has_number = any(char.isdigit() for char in password)

if has_number:
    print("✓ Contains a number")
else:
    print("✗ Missing a number")
```

```python
import string

has_special = any(char in string.punctuation for char in password)

if has_special:
    print("✓ Contains a special character")
else:
    print("✗ Missing a special character")
```

```python
import string

password = input("Enter a password: ")

score = 0

if len(password) >= 8:
    score += 1

if any(char.isupper() for char in password):
    score += 1

if any(char.islower() for char in password):
    score += 1

if any(char.isdigit() for char in password):
    score += 1

if any(char in string.punctuation for char in password):
    score += 1

print("Score:", score, "/ 5")
```

```python
if score <= 2:
    print("Strength: Weak")
elif score <= 4:
    print("Strength: Medium")
else:
    print("Strength: Strong")
```

```python
import string

password = input("Enter a password: ")

score = 0

# Length
if len(password) >= 8:
    score += 1

# Uppercase
if any(char.isupper() for char in password):
    score += 1

# Lowercase
if any(char.islower() for char in password):
    score += 1

# Number
if any(char.isdigit() for char in password):
    score += 1

# Special character
if any(char in string.punctuation for char in password):
    score += 1


print("\nPassword Analysis")
print("------------------")

print("Length:", len(password), "characters")
print("Score:", score, "/ 5")


if score <= 2:
    print("Strength: Weak")
elif score <= 4:
    print("Strength: Medium")
else:
    print("Strength: Strong")
```
